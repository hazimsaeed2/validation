## Readme

This notebook shows how to use databricks API to invoke the start of a job. 

In the context of the Coupons Personalization engine this will be done from an AWS Lambda function.  The API requires the JOB ID but that is created dinamically at each deployment, then, in order to be able to keep the Lambda code static we created an auxiliary method "get_job_id_by_name" to retrieve the Id given the job name (static)  

The API requires validation using OAuth credentials, for that purpose, credentials for the DEV and QA service principals are provided.

```python
from databricks.sdk import WorkspaceClient
```

```python
def get_job_id_by_name(w, job_name):
    jobs = w.jobs.list()   # returns a generator
    for job in jobs:
        if job.settings.name == job_name:
            return job.job_id
    return None
```

```python
# Get the job ID by job name (id may change, name shouldnt)
JOB_NAME = 'de_wf_ds__Personalization__pe_engine_orchestration'  # '<job-name>' this should have the name of the 

job_id = get_job_id_by_name(w, JOB_NAME)

print(job_id)
```

```python
w.jobs.run_now(job_id=job_id)
```
