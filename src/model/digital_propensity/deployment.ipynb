```python
source_environment = dbutils.widgets.get("source_environment").lower()
env = dbutils.widgets.get("env").lower()
```

```python
%run ../../config/utils
```

```python
import mlflow
mlflow.set_registry_uri('databricks-uc')

client = mlflow.tracking.MlflowClient()
```

## Trips model migration

```python
target_model = digital_trips_catalog
source_model = target_model.replace(env, source_environment)
```

```python
source_champion_version = client.get_model_version_by_alias(name=source_model, alias="champion").version

target_version = client.copy_model_version(
  f"models:/{source_model}@champion",
  target_model,
).version

client.set_registered_model_alias(target_model, "champion", target_version)
client.set_model_version_tag(
    name=target_model,
    version=target_version,
    key=f"{source_environment}_version",
    value=source_champion_version,
)
client.set_model_version_tag(
    name=target_model,
    version=target_version,
    key=f"{source_environment}_catalog_source",
    value=source_model,
)
```
